* Creating Open invoices & Aging data (Global B2B AR) for Poka-Yoke Agencies: 

```python
# importing our libraries 
import pandas as pd
import random
from datetime import datetime, timedelta

# Defining List of synthetic values as variables to key into the dataset 
fictional_companies = {
    "Alpha Logistics": ["California reg. Office", "76, 5th Ave, New York"], 
    "Beta Global": ["58, 5th Ave, New York", "Rotary Plaza, London, UK"], 
    "Delta Supply Chain": ["Carpentry Trade centre, Texas", ''], 
    "Omega Trading": ["123, Main St, Los Angeles", "456, Oak Ave, Chicago"], 
    "Vertex Industries": ["789, Pine Rd, Houston", "789, Pine Rd, Houston"],
    "Nexus Solutions": ["789, Pine Rd, Houston", "123, Main St, Los Angeles"],
    "Nexus Solutions New York": ["789, Pine Rd, Houston", "World Trade Centre, New York"],
    "Globpine cycles": ["Beeway U-bahnhoff, Stuttgart, Germany", "Bennington Square, London, UK"],
    "Globpine cycles France": ["Rue de la Paix, Paris, France"],
    "Globpine cycles Spain": ["Calle de Alcalá, Madrid, Spain"],
    "Nexaltech": ["Thatikallupuram Tech Park, Bangalore, India", ''],
    "Mahindra Logistics ltd.": ["Mahindra Towers, Mumbai, India", "Mahindra factory, Chennai, India"],
    "Mahindra Logistics ltd. UK": ["Mahindra Towers, London, UK", ''],
    "Cyclide Batteries": ["Kaiserstrasse 12, Berlin, Germany"],
    "Cyclide Batteries France": ["Kaiserstrasse 12, Paris, France", "Louvre Museum, Paris, France"],
    "Cyclide Batteries Spain": ["Kaiserstrasse 12, Madrid, Spain"]
    }

fictional_regions = ["NAM", "EMEA", "APAC", "LATAM"]

risk_levels = [
    "LOW-Lowest Risk", 
    "MED-Medium Risk", 
    "HIGH-High Risk"
    ]

reason_code = ["RT", "UN", "OV", "MI"]


# Defined the empty dataset
synthetic_AR_data = []

# Generating 500 rows of synthetic data by looping each row
for i in range(1, 501):

    doc_type = random.choice(["CM", "IN", "DN"]) # Defined the document types
    region = random.choice(fictional_regions)

    # Defining variables and calculations if doc_type is "IN" (i.e. an invoice line)
    if doc_type == "IN":
        doc_num = f"A{random.randint(1000000, 9999999)}"
        doc_date = datetime.now() - timedelta(days=random.randint(10, 120))
        term_days = random.choice([30,45,60,90])
        pay_terms = f"Net-{term_days}"
        due_date = doc_date + timedelta(days=term_days)
        days_late = (datetime.now() - due_date).days
        reason = None
        reas_desc = None
        inv_amt = round(random.randint(2500, 3500000),2)
        Curr_bal = 0.00
        d30_bal = 0.00
        d60_bal = 0.00
        d90_bal = 0.00
        d90_bal = 0.00
        d90plus_bal = 0.00 

        # Assigning the inv_amt to Aging buckets based on their respective due_dates
        if days_late < 0:
            Curr_bal = inv_amt
            aging_buckets = "Current"

        elif days_late <= 30:
            d30_bal = inv_amt
            aging_buckets = "0-30 days past due"
        
        elif days_late <= 60:
            d60_bal = inv_amt
            aging_buckets = "31-60 days past due"
        
        elif days_late <= 90:
            d90_bal = inv_amt
            aging_buckets = "61-90 days past due"
        
        elif days_late > 90:
            d90plus_bal = inv_amt
            aging_buckets = ">90 days past due"

        text = f"Invoice generated as on {doc_date.strftime('%m/%d/%Y')}"

    # Defining variables and calculations if doc_type is "CM" (i.e. a Credit memo)
    elif doc_type == "CM":
        doc_num = f"CM{random.randint(1000000, 9999999)}"
        doc_date = datetime.now() - timedelta(days=random.randint(5, 60))
        pay_terms = None
        due_date = None
        reason = None
        reas_desc = None
        inv_amt = round(random.randint(500, 150000),2)
        aging_buckets = None
        last_doc_num = random.randint(100000, 999999) #to refer to a random inv doc num in text
        text = random.choice([
            '',
            f"Credit Approved against inv A{last_doc_num}",
            "undelievred Goods",
            "No BOL copy",
            "NA",
            "N/A",
            (datetime.now() - timedelta(days=random.randint(1, 30))).strftime('%m/%d/%Y'),
            "Duplct Inv raised",
            "Dup invoice Raised"
            ])
    
    # Defining variables and calculations if doc_type is "DN" (i.e. a deduction line) 
    else:
        doc_num = f"DN{random.randint(100000, 999999)}"
        doc_date = datetime.now() - timedelta(days=random.randint(50, 120))
        pay_terms = None
        due_date = doc_date
        days_late = (datetime.now() - due_date).days
        reason = random.choice(reason_code)
        last_doc_num = random.randint(100000, 999999) #to refer to a random inv doc num in text

        # Defining the Reason descriptions for each reason code assigned to a DN
        if reason == "RT":
            reas_desc = "RETURNED"
            text = random.choice([
                f"Items returned agst A{last_doc_num}",
                "cust deducted against returns",
                "ITEM RETURNED",
                ''
            ])
        
        elif reason == "OV":
            reas_desc = "OVERCHARGED"
            text = random.choice([
                "Custmr overchgd",
                f"Customer paid twice on inv#{last_doc_num}",
                f"double pymt on inv#{last_doc_num}",
                '',
                "cust did not recv credit"
            ])
        
        elif reason == "UN":
            reas_desc = "Undelivered"
            text = random.choice([
                "Not delieved on time",
                "Products missing & POD not receivd",
                "PODmissing",
                '',
                "No POD"
            ])
        
        elif reason == "MI":
            reas_desc = "missing items"
            text = random.choice([
                "Incomplete shipmt recd",
                "shipmt not reCvd",
                '',
                "BOL missing, cant confirm shipment"
            ])

        inv_amt = round(random.randint(5000, 98000),2)
        Curr_bal = 0.00
        d30_bal = 0.00
        d60_bal = 0.00
        d90_bal = 0.00
        d90_bal = 0.00
        d90plus_bal = 0.00

        # Assigning the inv_amt to Aging buckets based on their respective due_dates
        if days_late < 0:
                Curr_bal = inv_amt
                aging_buckets = "Current"
    
        elif days_late <= 30:
            d30_bal = inv_amt
            aging_buckets = "0-30 days past due"
        
        elif days_late <= 60:
            d60_bal = inv_amt
            aging_buckets = "31-60 days past due"
        
        elif days_late <= 90:
            d90_bal = inv_amt
            aging_buckets = "61-90 days past due"
        
        elif days_late > 90:
            d90plus_bal = inv_amt
            aging_buckets = ">90 days past due"

    # Setting up the Company Codes based on regions assigned
    if region == "NAM":
        comp_code = random.choice([
            "NA22",
            "NA01",
            "NA08"
        ])
    elif region == "EMEA":
        comp_code = random.choice([
            "E012",
            "E016"
        ])
    elif region == "APAC":
        comp_code = random.choice([
            "AP21",
            "AP02"
        ])
    else:
        comp_code = "Other"

    # Defining the Customer name, Bill-to & Ship-to Address fields: 
    Select_Cust = random.choice(list(fictional_companies.keys())) 
    Cust_address = fictional_companies[Select_Cust]
    Select_BillTo = Cust_address[0]

    # Handling Edge cases: When Ship-To evaulates to '' or None 
    if len(Cust_address) > 1:
        Select_ShipTo = Cust_address[1]
        if Select_ShipTo == '':
            Select_ShipTp = "N/A"
    else:
        Select_ShipTo = None

    # Appending to the dataset as dictionary
    synthetic_AR_data.append({
        'Reporting Date': datetime.now().strftime('%m/%d/%Y %H:%M'),
        'Region': random.choice(fictional_regions),
        'Company': comp_code,
        'Customer Name': Select_Cust,
        'Bill-to Address': Select_BillTo,
        'Ship-to Address': Select_ShipTo,
        'Risk Category': random.choice(risk_levels),
        'Document Type': doc_type,
        'Document Number': doc_num,
        'Reason Code': reason,
        'Reason Desc.': reas_desc,
        'Payment Terms': pay_terms,
        'Document Date': doc_date,
        'Due Date': due_date,
        'Status': aging_buckets,
        'Text': text,
        'Current': Curr_bal,
        '0 - 30 days': d30_bal,
        '31 - 60 days': d60_bal,
        '61 - 90 days': d90_bal,
        '90+ days': d90plus_bal,
        'Grand Total': 0.0 # Will calculate later
    })

# Defining the dataframe with the appended dataset generated 
df_synthetic_AR_data = pd.DataFrame(synthetic_AR_data)


# Calculating the Grand Total (Total AR) and assigning to the dataframe 
df_synthetic_AR_data['Grand Total'] = round(
    df_synthetic_AR_data['Current'] 
    + df_synthetic_AR_data['0 - 30 days'] 
    + df_synthetic_AR_data['31 - 60 days'] 
    + df_synthetic_AR_data['61 - 90 days'] 
    + df_synthetic_AR_data['90+ days']
    ,2)

# Exporting to your local machine
df_synthetic_AR_data.to_excel("PokaYoke Agencies Global Aging raw data.xlsx", index=False)
print("Generated 100% safe, non-proprietary dummy data successfully!")
```
